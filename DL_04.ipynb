## Aim:
### To implement feed forward network for classification with synthetic data.

```python
import numpy as np

np.random.seed(3)
N = 300
X = np.random.randn(N,4)

true_W = np.array([[2,-1,0.5,1],
                   [-1,1,-0.5,-1],
                   [0.5,-0.5,1,0.5]])
true_b = np.array([0.5,-0.3,0.1])

logits = np.dot(X,true_W.T) + true_b
Y = np.argmax(logits + 0.5*np.random.randn(*logits.shape),axis=1)

Y_onehot = np.eye(3)[Y]

def relu(z): return np.maximum(0,z)
def relu_grad(z): return (z>0).astype(float)

def sigmoid(z): return 1/(1+np.exp(-z))
def sigmoid_grad(z): s=sigmoid(z); return s*(1-s)

def tanh(z): return np.tanh(z)
def tanh_grad(z): return 1-np.tanh(z)**2

def softmax(z):
    z = z - np.max(z,axis=1,keepdims=True)
    return np.exp(z)/np.sum(np.exp(z),axis=1,keepdims=True)

activation = "relu"

if activation=="relu":
    act = relu; act_grad = relu_grad
elif activation=="sigmoid":
    act = sigmoid; act_grad = sigmoid_grad
else:
    act = tanh; act_grad = tanh_grad

input_dim=4
hidden_dim=8
output_dim=3

W1 = 0.1*np.random.randn(input_dim,hidden_dim)
b1 = np.zeros((1,hidden_dim))
W2 = 0.1*np.random.randn(hidden_dim,output_dim)
b2 = np.zeros((1,output_dim))

lr = 0.05
epochs = 2000

for e in range(epochs):
    z1 = np.dot(X,W1) + b1
    h = act(z1)
    z2 = np.dot(h,W2) + b2
    y_hat = softmax(z2)

    loss = -np.mean(np.sum(Y_onehot*np.log(y_hat+1e-8),axis=1))

    dz2 = (y_hat - Y_onehot)/N
    dW2 = np.dot(h.T,dz2)
    db2 = np.sum(dz2,axis=0,keepdims=True)

    dh = np.dot(dz2,W2.T)
    dz1 = dh * act_grad(z1)
    dW1 = np.dot(X.T,dz1)
    db1 = np.sum(dz1,axis=0,keepdims=True)

    W1 -= lr*dW1
    b1 -= lr*db1
    W2 -= lr*dW2
    b2 -= lr*db2

    if (e+1)%400==0:
        preds = np.argmax(y_hat,axis=1)
        acc = np.mean(preds==Y)
        print(f"Epoch {e+1}  Loss={loss:.4f}  Acc={acc:.3f}")

preds = np.argmax(y_hat,axis=1)
print("\nFinal Accuracy:",np.mean(preds==Y))
```

## Result:
### Thus, the implementation of feed forward network for classification with synthetic data has been done successfully.

## Aim:
### To implement feed forward network for classification with a standard data set (like iris dataset, housing dataset etc.)

```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

iris = load_iris()
X = iris.data
y = iris.target

X = (X - X.mean(axis=0)) / X.std(axis=0)

n_samples, n_features = X.shape
n_classes = 3

Y = np.zeros((n_samples, n_classes))
for i in range(n_samples):
    Y[i, y[i]] = 1

X_train, X_test, Y_train, Y_test, y_train, y_test = train_test_split(
    X, Y, y, test_size=0.3, random_state=42
)

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def sigmoid_derivative(z):
    s = sigmoid(z)
    return s * (1 - s)

def softmax(z):
    z = z - np.max(z)
    exp_z = np.exp(z)
    return exp_z / np.sum(exp_z)

hidden_neurons = 8
learning_rate = 0.01
epochs = 2500

W1 = np.random.randn(n_features, hidden_neurons)
b1 = np.zeros(hidden_neurons)

W2 = np.random.randn(hidden_neurons, n_classes)
b2 = np.zeros(n_classes)

for epoch in range(epochs):
    total_loss = 0
    for i in range(len(X_train)):
        x = X_train[i]
        y_true = Y_train[i]

        z1 = np.dot(x, W1) + b1
        a1 = sigmoid(z1)

        z2 = np.dot(a1, W2) + b2
        y_pred = softmax(z2)

        loss = -np.sum(y_true * np.log(y_pred + 1e-8))
        total_loss += loss

        dL_dz2 = y_pred - y_true
        dW2 = np.outer(a1, dL_dz2)
        db2 = dL_dz2

        dL_da1 = np.dot(W2, dL_dz2)
        dL_dz1 = dL_da1 * sigmoid_derivative(z1)

        dW1 = np.outer(x, dL_dz1)
        db1 = dL_dz1

        W2 -= learning_rate * dW2
        b2 -= learning_rate * db2
        W1 -= learning_rate * dW1
        b1 -= learning_rate * db1

    if epoch % 500 == 0:
        print(f"Epoch {epoch}, Loss = {total_loss / len(X_train):.4f}")

correct = 0

for i in range(len(X_test)):
    z1 = np.dot(X_test[i], W1) + b1
    a1 = sigmoid(z1)

    z2 = np.dot(a1, W2) + b2
    y_pred = softmax(z2)

    if np.argmax(y_pred) == y_test[i]:
        correct += 1

accuracy = correct / len(X_test)
print("Test Accuracy using Sigmoid:", accuracy)
```

## Result:
### Thus, the implementation of feed forward network for classification with a standard data set has been done successfully.
