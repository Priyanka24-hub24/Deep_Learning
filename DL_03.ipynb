## Aim:
###    To implement Logic Function XOR using Feed forward network.

```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

def mse(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)

X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

Y = np.array([[0], [1], [1], [0]])

np.random.seed(1)
w_hidden = np.random.uniform(-1, 1, (2, 2))
b_hidden = np.random.uniform(-1, 1, (1, 2))

w_output = np.random.uniform(-1, 1, (2, 1))
b_output = np.random.uniform(-1, 1, (1, 1))

learning_rate = 0.5
epochs = 2000
checkpoints = [0, 500, 1000, 1500, 2000]

for epoch in range(epochs + 1):

    hidden_output = sigmoid(np.dot(X, w_hidden) + b_hidden)
    predicted_output = sigmoid(np.dot(hidden_output, w_output) + b_output)

    error = Y - predicted_output

    d_output = error * sigmoid_derivative(predicted_output)
    d_hidden = d_output.dot(w_output.T) * sigmoid_derivative(hidden_output)

    w_output += hidden_output.T.dot(d_output) * learning_rate
    b_output += np.sum(d_output, axis=0, keepdims=True) * learning_rate

    w_hidden += X.T.dot(d_hidden) * learning_rate
    b_hidden += np.sum(d_hidden, axis=0, keepdims=True) * learning_rate

    if epoch in checkpoints:
        print("\nEpoch:", epoch)
        print("Loss:", mse(Y, predicted_output))
        print("XOR Output:")
        for i in range(len(X)):
            print(X[i], "->", int(predicted_output[i][0] >= 0.5))
```

## Result:
## Thus the implementation of Function XOR using Feed forward network has been done successfully.
